```python
path = "/storage/vacoda/data/raw/api/single/comtrade_391211_2022.csv"

df = pd.read_csv(path)
group_cols = [
    "Period",
    "ReporterISO",
    "PartnerISO",
    "FlowCode",
    "CmdCode",
]

grouped_cif_fob = (
    df.groupby(group_cols, dropna=False)
    .agg(
        cif_value_sum=(
            "Cifvalue",
            lambda s: s.sum(min_count=1),
        ),
        fob_value_sum=(
            "Fobvalue",
            lambda s: s.sum(min_count=1),
        ),
        n_cif_rows=(
            "Cifvalue",
            lambda s: s.notna().sum(),
        ),
        n_fob_rows=(
            "Fobvalue",
            lambda s: s.notna().sum(),
        ),
        n_source_rows=(
            "Cifvalue",
            "size",
        ),
    )
    .reset_index()
)

groups_with_both = grouped_cif_fob[
    (grouped_cif_fob["n_cif_rows"] > 0)
    & (grouped_cif_fob["n_fob_rows"] > 0)
].copy()

print("Total grouped observations:", len(grouped_cif_fob))
print(
    "Groups containing both CIF and FOB:",
    len(groups_with_both),
)

groups_with_both.head(20)

```

```python
import pandas as pd

def inspect_group(example_row: pd.Series) -> pd.DataFrame:
    year = int(example_row["year"])
    reporter = str(example_row["reporter_iso"]).replace("'", "''")
    partner = str(example_row["partner_iso"]).replace("'", "''")
    flow = str(example_row["flow_code"]).replace("'", "''")
    target_cmd = str(example_row["target_cmd"]).replace("'", "''")

    return duckdb.sql(
        f"""
        SELECT
            hs_sync_input_row_id,

            year,
            reporter_iso,
            partner_iso,
            flow_code,
            target_cmd,

            partner2_iso,
            partner2_name,

            mot_code,
            mot_name,

            customs_code,
            customs_name,
            mos_code,

            source_hs,
            source_cmd,
            source_cmd_original,

            primary_value_original,
            primary_value_hs_synced,
            allocation_weight,

            cif_value_usd,
            fob_value_usd,

            net_weight_kg,

            is_reported,
            is_aggregate

        FROM read_parquet('{detail_path}')

        WHERE year = {year}
          AND reporter_iso = '{reporter}'
          AND partner_iso = '{partner}'
          AND flow_code = '{flow}'
          AND CAST(target_cmd AS VARCHAR) = '{target_cmd}'

        ORDER BY
            partner2_iso,
            mot_code,
            customs_code,
            mos_code,
            source_cmd
        """
    ).df()
```

```python
example_keys = duckdb.sql(
    f"""
    WITH groups AS (
        SELECT
            year,
            reporter_iso,
            partner_iso,
            flow_code,
            target_cmd,

            COUNT(*) AS n_reported_rows,

            COUNT(*) FILTER (
                WHERE CAST(mot_code AS VARCHAR) = '0'
            ) AS n_mot_total_rows,

            COUNT(*) FILTER (
                WHERE CAST(mot_code AS VARCHAR) <> '0'
            ) AS n_mot_detail_rows,

            COUNT(*) FILTER (
                WHERE partner2_iso = 'W00'
            ) AS n_partner2_total_rows,

            COUNT(*) FILTER (
                WHERE partner2_iso <> 'W00'
            ) AS n_partner2_detail_rows

        FROM read_parquet('{detail_path}')

        GROUP BY
            year,
            reporter_iso,
            partner_iso,
            flow_code,
            target_cmd
    ),

    classified AS (
        SELECT
            *,

            CASE
                WHEN n_reported_rows = 1
                    THEN 'single_row'

                WHEN n_mot_total_rows = 0
                 AND n_partner2_total_rows = 0
                    THEN 'partner2_detail_x_mot_detail'

                WHEN n_mot_total_rows = 0
                 AND n_partner2_detail_rows = 0
                    THEN 'partner2_total_x_mot_detail'

                WHEN n_mot_detail_rows = 0
                 AND n_partner2_total_rows = 0
                    THEN 'partner2_detail_x_mot_total'

                WHEN n_mot_detail_rows = 0
                 AND n_partner2_detail_rows = 0
                    THEN 'partner2_total_x_mot_total'

                ELSE 'unclassified'
            END AS structure_class

        FROM groups
    ),

    ranked AS (
        SELECT
            *,

            ROW_NUMBER() OVER (
                PARTITION BY structure_class
                ORDER BY
                    ABS(n_reported_rows - 3),
                    year DESC
            ) AS example_rank

        FROM classified

        WHERE structure_class IN (
            'single_row',
            'partner2_detail_x_mot_detail',
            'partner2_total_x_mot_detail',
            'partner2_detail_x_mot_total',
            'partner2_total_x_mot_total'
        )
    )

    SELECT
        structure_class,
        year,
        reporter_iso,
        partner_iso,
        flow_code,
        target_cmd,
        n_reported_rows

    FROM ranked

    WHERE example_rank = 1

    ORDER BY structure_class
    """
).df()

example_keys
```

```python
examples = {}

for _, example_row in example_keys.iterrows():
    structure_class = example_row["structure_class"]

    examples[structure_class] = inspect_group(example_row)

    print("=" * 100)
    print(structure_class)
    print(example_row.to_dict())
    display(examples[structure_class])
```
